```python
from google.colab import userdata
api_key=userdata.get('api_key')
```

```python
from openai import OpenAI
import base64
from PIL import Image
from IPython.display import display

client = OpenAI(
    api_key=api_key,
    base_url="https://nexusapi.navigatelabs.ai"
)

# Generate image
result = client.images.generate(
    model="nano-banana",
    prompt="A majestic snow leopard standing on a mountain peak at sunrise"
)

# Decode the base64 image
image_base64 = result.data[0].b64_json
image_bytes = base64.b64decode(image_base64)

# Save the image
image_path = "canva.png"
with open(image_path, "wb") as f:
    f.write(image_bytes)

print(f"Image saved as {image_path}")

# Display the image in Colab
try:
    img = Image.open(image_path)
    display(img)
except FileNotFoundError:
    print(f"{image_path} not found. Please make sure the image generation was successful.")
```
