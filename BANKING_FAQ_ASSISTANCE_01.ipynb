Step 1:static context

```python
SYSTEM_CONTEXT = """
You are a banking FAQ assistant.

Rules:
- Answer the user's question using the provided context.
- Do not invent banking policies.
- If the answer is not available in the context, say:
  "I don't have enough information to answer that."
- Never ask for OTP, PIN, CVV, or password.
"""
```

Step 2 :External context

```python
BANKING_POLICY = """
Banking Refund Policy:
- Refund requests are allowed within 7 days of a transaction.
- The transaction must be eligible for reversal.
- Completed and settled transactions may not be refundable.
- Subscription-based services are non-refundable.
"""
```

Step 3:Dynamic context

```python
user_profile = {
    "account_type": "Savings Account",
    "customer_type": "Premium",
    "transaction_days_ago": 3
}

user_query = "Can I get a refund for my transaction?"
```

Step 4:Context assembly


```python
final_prompt = f"""
{SYSTEM_CONTEXT}

Banking Policy:
{BANKING_POLICY}

User Profile:
- Account Type: {user_profile['account_type']}
- Customer Type: {user_profile['customer_type']}
- Transaction: {user_profile['transaction_days_ago']} days ago

User Question:
{user_query}
"""

print(final_prompt)
```

```python
from google.colab import userdata

# Access the API key stored in Colab Secrets
api_key = userdata.get('api_key')

# Now, use this variable when initializing your OpenAI client
client = OpenAI(api_key=api_key, base_url="https://nexusapi.navigatelabs.ai")

print("API key loaded successfully from Colab Secrets.")
```

Step 5:Call the model

```python
from openai import OpenAI

client = OpenAI(
    api_key=api_key,
    base_url="https://nexusapi.navigatelabs.ai"
)

response = client.chat.completions.create(
    model="gpt-4.1-nano",
    messages=[
        {
            "role": "system",
            "content": SYSTEM_CONTEXT
        },
        {
            "role": "user",
            "content": final_prompt
        }
    ]
)

print(response.choices[0].message.content)
```
